```python
    def isValid(self, s: str) -> bool:
        if not s:
            return False
        check = []
        for ch in s:
            if ch in '({[':
                check.append(ch)
            elif ch == ')':
                if not check or check[-1] != '(':
                    return False
                check.pop()
            elif ch == ']':
                if not check or check[-1] != '[':
                    return False
                check.pop()
            elif ch == '}':
                if not check or check[-1] != '{':
                    return False
                check.pop()

        return not check
```
